# Finalists Prediction Notebook

This notebook trains conservative classification models to predict World Cup finalists (top-9) using historical World Cup data. The notebook is leakage-aware (it removes direct target-derived features) and follows nested cross-validation for honest evaluation.

Outputs:
- `outputs/finalists/finalists_top_all.csv` — ensemble probabilities for all modeled teams
- `outputs/finalists/finalists_top9.csv` — top-9 predicted finalists (ensemble)

Run cells from top to bottom. Heavy operations (nested CV) may take time depending on environment.
